<h3>

```python
# Variables a las que se les va a realizar un conteo de valores del df
variables = ['estado', 'sexo', 'tipo_de_contagio', 'nombre_del_pais', 'tipo_de_recuperacion', 'pertenencia_etnica', 'nombre_del_grupo_etnico']

# Iteración para validar el conteo de valores de cada variable
for var in variables:
    if var in data_covid.columns:
        print(f'{data_covid[var].value_counts(dropna=False)}\n')
```

<h3>
    Limpieza de datos de la columna estado
</h3>

```python
# Se define una lista con la cantidad de valores únicos de la columna estado
list_states = list(data_covid['estado'].unique())

# Lista vacía que contendrá los valores únicos en mayúscula
upper_states = []

# Iteración de los estados, transformación a mayúscula y agregación a la lista vacía
for state in list_states:
    upper = str(state).upper()
    if upper == 'NAN':
        upper_states.append('ASINTOMATICO')
    else:
        upper_states.append(upper)

# Se crea el diccionario entre la lista de estados y su respectiva transfomación a mayúsculas
states_dictionary = dict(zip(list_states, upper_states))

# Se realiza el mapeo del diccionario en el df
data_covid['estado'] = data_covid['estado'].map(states_dictionary)

# Se imprime el conteo de valores
data_covid['estado'].value_counts(dropna=False)
```

```python
# Se crea una lista de los valores únicos de la columna nombre_del_país
countries_list = data_covid['nombre_del_pais'].unique()

# Se crea una lista con los nombres de los paises a corregir
wrong_countries = ['ARABIA SAUDÍ', 'BELARUSIA', 'ESTADOS UNIDOS DE AMÉRICA', 'LIBANO', 'MEXICO', 'PANAMA', 'PERU', 'REINO UNIDO DE GRAN BRETAÑA E IRLANDA DEL NORTE', 'REPÚBLICA DOCIMINCANA', 'SANTA HELENA', 'TABATINGA', 'VENEUELA']

# Se crea una lista con los nombres de los paises correctos
right_countries = ['ARABIA SAUDITA', 'BIELORRUSIA', 'ESTADOS UNIDOS', 'LÍBANO', 'MÉXICO', 'PANAMÁ', 'PERÚ', 'REINO UNIDO', 'REPÚBLICA DOMINICANA', 'REINO UNIDO', 'BRASIL', 'VENEZUELA']

# Se crea la lista vacía que contendrá los valores correctos
new_countries_list = []

# Se realiza la iteración de las listas para llenar la lista que contendrá los valores correctos
for country in countries_list:
    if country in wrong_countries:
        new_countries_list.append(right_countries[wrong_countries.index(country)])
    else:
        new_countries_list.append(country)

# Se crea el diccionario entre la lista de paises y su respectiva transfomación
country_dictionary = dict(zip(countries_list, new_countries_list))

# Se realiza el mapeo del diccionario en el df
data_covid['nombre_del_pais'] = data_covid['nombre_del_pais'].map(country_dictionary)

# Se imprime el conteo de valores
data_covid['nombre_del_pais'].value_counts(dropna=False)
```

<h3>
    Valores atípicos
</h3>

```python
# Calcula el Primer Cuartil (Q1): el valor por debajo del cual se encuentra el 25% de las edades más bajas
Q1 = data_covid_clean['edad'].quantile(0.25)

# Calcula el Tercer Cuartil (Q3): el valor por debajo del cual se encuentra el 75% de las edades
Q3 = data_covid_clean['edad'].quantile(0.75)

# Calcula el Rango Intercuartílico (IQR): la diferencia entre Q3 y Q1. Representa el 50% central de los datos
IQR = Q3 - Q1

# Establece el límite inferior. Toda edad menor a este número se considerará un dato atípico (excesivamente joven/error)
lower_bound = Q1 - 1.5 * IQR

# Establece el límite superior. Toda edad mayor a este número se considerará un dato atípico (excesivamente mayor/error)
upper_bound = Q3 + 1.5 * IQR

# Filtra el df para aislar las filas donde la edad es menor al límite inferior o mayor al límite superior
outliers = data_covid_clean[(data_covid_clean['edad'] < lower_bound) | (data_covid_clean['edad'] > upper_bound)]

print(f'Número de datos atípicos de la columna edad: {len(outliers)}\n')

print(f'Porcentaje de datos atípicos de la columna edad: {(len(outliers) / len(data_covid_clean)) * 100:.2f}%\n')

print('Datos atípicos de la columna edad:\n')

# Muestra un resumen estadístico (media, min, max, etc.) de la columna edad
data_covid_clean['edad'].describe().round(2)
```

```python
# Aplica la técnica de "Recorte" (Clipping). En lugar de borrar las filas atípicas, 
# reemplaza las edades extremas con el valor exacto de los límites que calculamos anteriormente
data_covid_clean['edad'] = data_covid_clean['edad'].clip(lower=lower_bound, upper=upper_bound)

# Muestra el resumen estadístico después del recorte para verificar que los valores min y max ya no superan los límites.
data_covid_clean['edad'].describe().round(2)
```

    Transformación de la edad a años

```python
# Se transforman las edades en años que se almacenará en la columna edad_años
data_covid_clean['edad_años'] = data_covid_clean.apply(lambda row: 
    row['edad'] if row['unidad_de_medida_de_edad'] == 1 # Años
    else row['edad'] / 12 if row['unidad_de_medida_de_edad'] == 2 # Meses
    else row['edad'] / 365 if row['unidad_de_medida_de_edad'] == 3 # Días
    else row['edad'], axis=1 
)

#Se eliminan las columnas edad y unidad_de_medida_de_edad
del data_covid_clean['edad']
del data_covid_clean['unidad_de_medida_de_edad']

#Validamo el df con la edad estandarizada
data_covid_clean.head()
```

<h3>

```python
# Se define la lista con el nombre de las columnas que tienen información de fechas
columns_date = ['fecha_reporte_web', 'fecha_de_notificacion', 'fecha_de_inicio_de_sintomas', 'fecha_de_muerte', 'fecha_de_diagnostico', 'fecha_de_recuperacion']

# Transformación de las columnas a formato datetime
for column in columns_date:
    if column in data_covid_clean.columns:
        data_covid_clean[column] = pd.to_datetime(data_covid_clean[column], errors='coerce')

data_covid_clean.head()
```
